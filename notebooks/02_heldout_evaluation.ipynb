# Held-out evaluation
Run the cells in order with the project environment. Outputs are written to results/.

```python
from pathlib import Path
import sys
import subprocess
cwd = Path.cwd()
root = next(p for p in [cwd,*cwd.parents] if (p/'src/pilot.py').is_file() and (p/'data/provenance.json').is_file())
print(root)
```

```python
# Install the dependencies in this kernel if needed.
# subprocess.check_call([sys.executable, "-m", "pip", "install", "-r", str(root / "requirements.txt")])
```

```python
subprocess.run([sys.executable,'-u',str(root/'src/holdout.py')],check=True)
```

```python
import pandas as pd
out = root/'results/holdout'
display(pd.read_csv(out/'summary.csv').round(4))
display(pd.read_csv(out/'paired_ci.csv').round(4))
```
